```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Carregar a imagem em escala de cinza
caminho_imagem = "pele-santos.jpg"  # Substitua pelo caminho da imagem
imagem_original = cv2.imread(caminho_imagem, cv2.IMREAD_GRAYSCALE)
```

```python
# Equalização do histograma
imagem_equalizada = cv2.equalizeHist(imagem_original)

# Criar os histogramas
histograma_original = cv2.calcHist([imagem_original], [0], None, [256], [0, 256])
histograma_equalizado = cv2.calcHist([imagem_equalizada], [0], None, [256], [0, 256])
```

```python
# Criar a visualização
fig, eixos = plt.subplots(2, 2, figsize=(12, 8))

# Exibir a imagem original
eixos[0, 0].imshow(imagem_original, cmap="gray")
eixos[0, 0].set_title("Imagem Original")
eixos[0, 0].axis("off")

# Exibir histograma da imagem original
eixos[0, 1].plot(histograma_original, color="black")
eixos[0, 1].set_title("Histograma Original")

# Exibir a imagem equalizada
eixos[1, 0].imshow(imagem_equalizada, cmap="gray")
eixos[1, 0].set_title("Imagem Equalizada")
eixos[1, 0].axis("off")

# Exibir histograma da imagem equalizada
eixos[1, 1].plot(histograma_equalizado, color="black")
eixos[1, 1].set_title("Histograma Equalizado")

# Ajustar layout e exibir
plt.tight_layout()
plt.show()
```
