#### 🧠 Parte Teórica

**Transformação de cores** é o processo de conversão de uma imagem de um espaço de cor para outro. Três espaços de cor comuns são:

- **RGB (Red, Green, Blue)**: padrão em câmeras e monitores, representa a imagem como a combinação de luz vermelha, verde e azul.
- **Grayscale (tons de cinza)**: converte uma imagem colorida para uma versão monocromática baseada na luminância.
- **HSV (Hue, Saturation, Value)**: separa a cor (matiz), a intensidade (saturação) e o brilho (valor), sendo útil para segmentação de cores.

Essas transformações são amplamente utilizadas em aplicações como segmentação de imagem, realce, análise de objetos e visão computacional.

#### 🔧 Parte Prática – Python + OpenCV

```python
# Instalar OpenCV se necessário
# !pip install opencv-python matplotlib

import cv2
import matplotlib.pyplot as plt
```

```python
# Carrega a imagem no formato RGB
imagem_original = cv2.imread('Pele.jpg')  # Substitua pelo caminho da sua imagem
imagem_rgb = cv2.cvtColor(imagem_original, cv2.COLOR_BGR2RGB)

# Converte para tons de cinza
imagem_cinza = cv2.cvtColor(imagem_original, cv2.COLOR_BGR2GRAY)

# Converte para o espaço de cor HSV
imagem_hsv = cv2.cvtColor(imagem_original, cv2.COLOR_BGR2HSV)

# Visualiza as imagens lado a lado
figura, eixos = plt.subplots(1, 3, figsize=(15, 5))

eixos[0].imshow(imagem_rgb)
eixos[0].set_title('Imagem em RGB')
eixos[0].axis('off')

eixos[1].imshow(imagem_cinza, cmap='gray')
eixos[1].set_title('Imagem em Tons de Cinza')
eixos[1].axis('off')

eixos[2].imshow(cv2.cvtColor(imagem_hsv, cv2.COLOR_HSV2RGB))
eixos[2].set_title('Imagem em HSV (convertida para RGB)')
eixos[2].axis('off')

plt.tight_layout()
plt.show()
```
